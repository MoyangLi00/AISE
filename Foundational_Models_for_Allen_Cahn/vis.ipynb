```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Visualize the training data
train_data_list = {}
basedir = '../Foundational_Models_for_Allen_Cahn/dataset/train'
train_datafile_list = [f for f in os.listdir(basedir) if f!="train_data.npz"]
train_datafile_list.sort()
print(train_datafile_list)

for i, train_datafile in enumerate(train_datafile_list):
    train_data_list[i] = np.load(os.path.join(basedir, train_datafile), allow_pickle=True)

# Plot the training data
grid = np.linspace(-1, 1, 128)
fig, axs = plt.subplots(3, 3, figsize=(30, 15))
for i, train_data in train_data_list.items():
    ax = axs[i // 3, i % 3]
    ax.plot(grid, train_data[0, 0, :], label = "$u(t = 0.00)$")
    ax.plot(grid, train_data[0, 1, :], label = "$u(t = 2.5e-5)$")
    ax.plot(grid, train_data[0, 2, :], label = "$u(t = 5e-5)$")
    ax.plot(grid, train_data[0, 3, :], label = "$u(t = 7.5e-5)$")
    ax.plot(grid, train_data[0, 4, :], label = "$u(t = 1e-4)$")
    ax.set_xlabel("x")
    ax.set_ylabel("u")
    ax.set_title(train_datafile_list[i])
    ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Visualize the test data
test_data_list = {}
basedir = '../Foundational_Models_for_Allen_Cahn/dataset/test'
test_datafile_list = [f for f in os.listdir(basedir) if f!="test_data.npz"]
test_datafile_list.sort()

for i, test_datafile in enumerate(test_datafile_list):
    test_data_list[i] = np.load(os.path.join(basedir, test_datafile))

# Plot the test data
grid = np.linspace(-1, 1, 128)
fig, axs = plt.subplots(5, 3, figsize=(30, 25))
for i, test_data in test_data_list.items():
    ax = axs[i // 3, i % 3]
    ax.plot(grid, test_data[0, 0, :], label = "$u(t = 0.00)$")
    ax.plot(grid, test_data[0, 1, :], label = "$u(t = 2.5e-5)$")
    ax.plot(grid, test_data[0, 2, :], label = "$u(t = 5e-5)$")
    ax.plot(grid, test_data[0, 3, :], label = "$u(t = 7.5e-5)$")
    ax.plot(grid, test_data[0, 4, :], label = "$u(t = 1e-4)$")
    ax.set_xlabel("x")
    ax.set_ylabel("u")
    ax.set_title(test_datafile_list[i])
    ax.legend()
plt.tight_layout()
plt.show()
```
